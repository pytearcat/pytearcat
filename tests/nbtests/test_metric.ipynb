```python
import pytearcat as pt
from itertools import product
```

```python
t,x,y,z = pt.coords('t,x,y,z')
g = pt.metric('ds2 = -dt**2 + dx**2 + dy**2 + dz**2')
```

```python
# Test 1

Left = g("_i,_j") * g("^i,^k")
Right = pt.kdelta()("^i,_j") # Identity matrix

for i, in product(range(4), repeat=1):

    left = Left.elements[i]
    right = Right.elements[i]
    assert left == right, f"Test 1: Raising and lowering indices via metric tensor not working correctly."
```

```python
# Some functions to be used in the tests
phi = pt.fun('phi','t,x,y,z')
Ax = pt.fun('Ax','t,x,y,z')
Ay = pt.fun('Ay','t,x,y,z')
Az = pt.fun('Az','t,x,y,z')
```

```python
# Test 2
# Define an arbitrary Tensor of rank 1

A = pt.ten('A',1)
A.assign([-1 * phi, 2 * Ax, 3 * Ay, 4 * Az],'^')
A.complete('^')

# Define the left and right-hand sides of the equation 
Left = g("^i,^j") * A("_i")
Right = A("^j")

Left = g("^i,^j") * A("_i")
Right = A("^j")

for i, in product(range(4), repeat=1):

    left = Left.elements[i]
    right = Right.elements[i]
    assert left == right, f"Test 2: Raising and lowering indices via metric tensor not working correctly."
```

```python
# Test 3
# Define an arbitrary Tensor of rank 2
A = pt.ten('A',2)
A.assign([[-1 * phi ** 2, 2 * phi * Ax, 3 * phi * Ay, 4 * phi * Az],
          [-1 * phi * Ax, 2 * Ax * Ax, -2 * Ax * Ay, -4 * Ax * Az],
          [-1 * phi * Ay, 7 * Ay * Ax, 3 * Ay * Ay, 18 * Ay * Az],
          [-1 * phi * Az, -5 * Az * Ax, 3 * Az * Ay, -5 * Az * Az]
          ],'_,_')
A.complete('_,_')
for i,j in product(range(4), repeat=2):

    left = Left.elements[i]
    right = Right.elements[i]
    assert left == right, f"Test 3: Raising and lowering indices via metric tensor not working correctly."
```
